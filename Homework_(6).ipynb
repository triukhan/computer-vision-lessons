```python
import cv2
import math
import numpy as np
from sklearn.utils import shuffle
import os
from matplotlib import pyplot as plt
from google.colab import drive

plt.rcParams['figure.figsize'] = [15, 10]
drive.mount('/content/drive')
folder = '/content/drive/MyDrive/subset_homework/'
```

```python
frames_0 = os.listdir(os.path.join(folder, 'class_id_0'))
images_0 = [cv2.imread(os.path.join(folder, 'class_id_0', f), cv2.IMREAD_UNCHANGED) for f in frames_0]
labels_0 = [0] * len(images_0)

frames_1 = os.listdir(os.path.join(folder, 'class_id_1'))
images_1 = [cv2.imread(os.path.join(folder, 'class_id_1', f), cv2.IMREAD_UNCHANGED) for f in frames_1]
labels_1 = [1] * len(images_1)
```

```python
images = images_0 + images_1
labels = labels_0 + labels_1
images, labels = shuffle(images, labels)
images, labels = np.array(images), np.array(labels)

pixels = np.array([img.flatten() for img in images]) / 255
```

```python
import tensorflow as tf
from tensorflow.keras import Model
from tensorflow.keras.layers import Dense, Input

inputs = Input(shape=(pixels.shape[1],))
x = Dense(64, activation='gelu')(inputs)
x = Dense(32, activation='gelu')(x)
x = Dense(16, activation='gelu')(x)
x = Dense(8, activation='gelu')(x)
outputs = Dense(1, activation='linear')(x)
model = Model(inputs, outputs)

model.summary()
model.compile(optimizer = 'adam', loss='mean_squared_error')
```

```python
model.fit(pixels, labels, epochs=100, batch_size=32)
```

```python
predictions = model.predict(pixels).squeeze()
predictions = predictions > 0.5
print(len([pred for pred in  predictions if pred]))

correct = 0

for prediction, label in zip(predictions, labels):
  if prediction == label:
    correct+=1

print(correct / len(labels))
```

For now its the best result. I added layers with some quantity of neurons and non-linear activation. The result at start was ~0.96, for now 0.999, actualy after adding just:
